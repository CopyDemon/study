```python
"""
In this practice we will list some port let python to check
if the port is open and running then we will close it one by one
"""
```

```python
import socket
import subprocess

ports = [8000, 8001, 8002, 8003, 8004, 8005, 8006, 8007, 8008, 8009, 8010, 8011, 8012, 8013, 8014, 8015, 8016, 8017, 8018, 8019, 49999, 50000, 50001, 50002, 50003,50004,50005,50006,50007]

# Loop through ports and check port is running then shutdown
def shutdown_ports(ports):
    for port in ports:
        # Check if port is in use
        port_in_use = check_port_in_use(port)
        
        print(f'Checking port: {port}......')
        
        # if in use
        if(port_in_use):
            print(f'port {port} is in use, getting PID......')
            # Check port process PID
            current_port_pid = find_process_using_port(port)
            print(f'port {port} PID is : {current_port_pid} and shutting it down now......')
            kill_process(current_port_pid)
            print(f'port{port} with PID: {current_port_pid} is already shutdown!')
        else:
            print(f'No process running on port: {port}')
        
        
# Check if the port is in use
def check_port_in_use(port):
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        return s.connect_ex(("localhost", port)) == 0
    
# Find port running process ID
def find_process_using_port(port):
    result = subprocess.run(['lsof', '-i', f':{port}'], capture_output=True, text=True)
    if(result.returncode == 0):
        return result.stdout.splitlines()[1].split()[1] # get process id by modify output string
    return None

def kill_process(pid):
    subprocess.run(['kill', pid])
        

shutdown_ports(ports)
```
