```python
"""
In this file I'm learning thread in python,
Reason I need to learn is because when a while loop is running in python it will block rest of code below it.
The solution is to use a different thread to process the while loop and other code below won't be blocked.
"""
```

```python
import time
# in here is the code will block the process
number = 0

# Here add while loop and loop forwever
"""
you will see the process is blocked, because while loop is running forever, 
and you won't see the print message:process is continued, not blocked
"""
while True:
    number += 1
    print(number)
    print("you cant see the message below, because the while loop blocked process")
    time.sleep(1)

# print process is ok
print("process is continued, not blocked")
```

```python
"""
In here we use thread to create a new thread for while loop then the while loop won't block the process
"""
```

```python
from threading import Thread
import time
variable = 0

def loop():
    global variable
    last_value = variable
    first_time_print = True
    while True:
        if variable != last_value:
            print("find different")
            last_value = variable
        elif first_time_print:
            print("Loop started!")
            first_time_print = False
        time.sleep(1)

# Here start thread target is the loop function
monitor_thread = Thread(target=loop)
monitor_thread.start()
```
